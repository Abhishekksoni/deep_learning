```python
torch.manual_seed(1234)
one_random = torch.rand(2,3)
two_random = torch.rand(2,3)

one_random, two_random, one_random.device, two_random.device
```

```python
one_random.to(device = 'cuda')
```

```python
two_random.to(device = 'cuda')
```

```python
transpose_martix = two_random.mT

matrix_mul = one_random @ transpose_martix
matrix_mul
```

```python
max = torch.max(matrix_mul)
min = torch.min(matrix_mul)
max, min
```

```python
max_index = torch.argmax(matrix_mul)
min_index = torch.argmin(matrix_mul)
max_index, min_index
```
